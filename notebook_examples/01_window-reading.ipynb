# Read and Visualize a 2D Microstructure Window
## About this Notebook:
This notebook is meant to be run on a standard computer. Examples of windows can be found in the directory 'example_data/window_dataset.npz'. This notebook will show how to load and plot windows before we compute the volume exchange in 02_volume-exchange.ipynb

There are 5 different classes of microstructures in the example data. There are already 2 independent sets of size 1200 consisting of sampled windows for each class.

```python
# ======== Library Imports ========
import numpy as np
import matplotlib.pyplot as plt
import sys
sys.path.append("../example_data")
```

```python
# Load windows
wins = np.load("../example_data/window_dataset.npz")
Classes1 = ["EI1", "EII1", "BII1", "RI1", "BI1"]
Classes2 = ["EI2", "EII2", "BII2", "RI2", "BI2"]
```

```python
# Plot example window from each class:
fig, axes = plt.subplots(figsize=(9,12), nrows=5,ncols=2,tight_layout=True, dpi=150)
for i in range(5):
    c1 = Classes1[i]
    c2 = Classes2[i]
    win1 = wins[c1][550] 
    win2 = wins[c2][550]
    
    axes[i][0].imshow(win1, cmap='tab20c')
    axes[i][0].set_title(c1)
    axes[i][0].axis('off')
    
    axes[i][1].imshow(win2, cmap='tab20c')
    axes[i][1].set_title(c2)
    axes[i][1].axis('off')
```
